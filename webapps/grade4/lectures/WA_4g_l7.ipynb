## Część 2 - Instrukcje warunkowe 🔀

Dotąd program wykonywał wszystko po kolei. Teraz nauczymy go **podejmować decyzje**: *jeżeli* warunek jest spełniony, zrób jedno; *w przeciwnym razie* zrób coś innego.

### 2.1 Wartości logiczne i porównania

Decyzje opierają się na pytaniach, na które odpowiedź brzmi `True` lub `False`. Służą do tego **operatory porównania**:

| Operator | Znaczenie |
|----------|-----------|
| `==` | równe (uwaga: podwójny znak!) |
| `!=` | różne |
| `>` &nbsp; `<` | większe / mniejsze |
| `>=` &nbsp; `<=` | większe-równe / mniejsze-równe |


````python
print(5 > 3)       # True
print(5 == 3)      # False
print(5 != 3)      # True
print(10 <= 10)    # True
````

Warunki można łączyć operatorami logicznymi **`and`**, **`or`**, **`not`**:
- `and` - prawda, gdy **oba** warunki prawdziwe,
- `or` - prawda, gdy **przynajmniej jeden** jest prawdziwy,
- `not` - odwraca wartość.


````python
age = 20
has_ticket = True

print(age >= 18 and has_ticket)   # True  - adult AND has a ticket
print(age < 18 or has_ticket)     # True  - at least one is true
print(not has_ticket)             # False
````

### 2.2 `if` - najprostsza decyzja

Składnia:

```python
if condition:
    # this runs only when condition is True
    ...
```

Dwie **kluczowe** zasady Pythona:
1. Po warunku stawiamy **dwukropek `:`**.
2. Kod „wewnątrz" `if` jest **wcięty** (wcięcie = 4 spacje). Wcięcie to nie estetyka - to ono mówi Pythonowi, co należy do bloku.


````python
temperature = 30

if temperature > 25:
    print("It is hot today.")
    print("Remember to drink water!")

print("This line always runs - it is not indented.")
````

### 2.3 `if / else` - dwie drogi

`else` („w przeciwnym razie") wykonuje się, gdy warunek jest `False`.


````python
number = 7

if number % 2 == 0:
    print(number, "is even.")
else:
    print(number, "is odd.")
````

### 2.4 `if / elif / else` - wiele możliwości

Gdy wariantów jest więcej, używamy `elif` (skrót od *else if*). Python sprawdza warunki **po kolei** i wchodzi do **pierwszego** spełnionego; resztę pomija.


````python
mark = 4   # a school mark from 1 to 6

if mark == 6:
    print("Outstanding!")
elif mark == 5:
    print("Very good.")
elif mark == 4:
    print("Good.")
elif mark == 3:
    print("Satisfactory.")
else:
    print("You need to work a bit more.")
````

## Część 3 - Pętle 🔁

Pętle pozwalają **powtarzać** kod bez przepisywania go wiele razy. To jedna z najważniejszych rzeczy w programowaniu - komputer nigdy się nie nudzi.

### 3.1 Pętla `for` i funkcja `range()`

`for` powtarza blok kodu dla każdej wartości z jakiejś sekwencji. Najczęściej używamy jej z `range()`, które generuje kolejne liczby.

- `range(5)` → `0, 1, 2, 3, 4` (pięć liczb, **od zera**, bez piątki)
- `range(1, 6)` → `1, 2, 3, 4, 5` (od 1 do 6, bez 6)
- `range(0, 10, 2)` → `0, 2, 4, 6, 8` (co 2)


````python
# Repeat something a fixed number of times
for i in range(5):
    print("Iteration number", i)
````

````python
# Counting from 1 to 5
for number in range(1, 6):
    print(number)

print("Lift-off!")
````

Zmienna pętli (tu `i` albo `number`) w każdym obiegu przyjmuje kolejną wartość. Możemy jej użyć w obliczeniach - np. tabliczka mnożenia:

````python
# Multiplication table for 7
n = 7
for i in range(1, 11):
    print(n, "x", i, "=", n * i)
````

`for` potrafi też chodzić bezpośrednio po elementach listy lub po znakach tekstu - bez `range()`.

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("I like", fruit)
````

### 3.2 Pętla `while`

`while` powtarza blok **dopóki** warunek jest prawdziwy. Używamy jej, gdy **nie wiemy z góry**, ile razy trzeba powtórzyć.

⚠️ **Uwaga:** wewnątrz pętli musi dziać się coś, co w końcu uczyni warunek fałszywym - inaczej powstanie **pętla nieskończona**. (Gdyby się zapętliło: *Runtime → Interrupt execution* albo przycisk ⏹.)


````python
# Count down from 5 to 1
count = 5

while count > 0:
    print(count)
    count = count - 1     # this line eventually stops the loop

print("Done!")
````

### 3.3 `break` i `continue`

Dwa słowa, które sterują pętlą od środka:
- **`break`** - natychmiast **przerywa** całą pętlę,
- **`continue`** - **pomija** resztę bieżącego obiegu i przechodzi do następnego.


````python
# break - stop as soon as we find the number 3
for i in range(1, 10):
    if i == 3:
        print("Found 3 - stopping the loop.")
        break
    print("Checking", i)
````

````python
# continue - skip even numbers, print only odd ones
for i in range(1, 11):
    if i % 2 == 0:
        continue          # skip the rest for even numbers
    print(i, "is odd")
````

### 3.4 Przykładowy program - suma liczb + FizzBuzz

**(a)** Suma liczb od 1 do `n` z pętlą `for`:


````python
# Sum of numbers from 1 to n
n = 10
total = 0

for i in range(1, n + 1):
    total = total + i

print("The sum from 1 to", n, "is", total)
````

**(b)** Klasyczne zadanie **FizzBuzz**:
dla liczb 1–15 wypisz `Fizz` jeśli liczba dzieli się przez 3, `Buzz` jeśli przez 5, `FizzBuzz` jeśli przez oba, a w pozostałych przypadkach samą liczbę.


````python
# FizzBuzz for numbers 1 to 15
for i in range(1, 16):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````

## 🎮 Mini-projekt - gra „zgadnij liczbę"

Teraz łączymy **wszystko**: zmienne, `input`, warunki, pętlę `while` oraz `break`.

Komputer losuje liczbę od 1 do 20, a gracz zgaduje. Po każdej próbie program podpowiada *„za mało"* lub *„za dużo"*, aż do trafienia.


````python
import random   # a module that lets us draw random numbers

secret = random.randint(1, 20)   # the number to guess, from 1 to 20
attempts = 0

print("I am thinking of a number between 1 and 20.")

while True:                      # loop "forever" - we leave it with break
    guess = int(input("Your guess: "))
    attempts = attempts + 1

    if guess < secret:
        print("Too low. Try again.")
    elif guess > secret:
        print("Too high. Try again.")
    else:
        print("Well done! You got it in", attempts, "attempt(s).")
        break                    # correct guess - leave the loop
````
