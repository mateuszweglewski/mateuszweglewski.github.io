## Część 1 — Czym jest lista 📋

**Lista** to uporządkowany zbiór wielu wartości trzymanych w **jednej** zmiennej. Zamiast pięciu osobnych zmiennych na pięć ocen — jedna lista.

Listę tworzymy w **nawiasach kwadratowych** `[ ]`, elementy oddzielamy przecinkami.


````python
# A list of marks and a list of names
marks = [5, 3, 4, 6, 2]
names = ["Anna", "Tom", "Kate", "John"]
mixed = [1, "two", 3.0, True]   # a list may hold different types

print(marks)
print(names)
print("Number of marks:", len(marks))   # len() gives the length
````

### 1.1 Indeksowanie — dostęp do elementu

Każdy element ma swój **indeks (numer pozycji)**. Uwaga: **liczymy od zera!**

```
  names = [ "Anna", "Tom", "Kate", "John" ]
  index      0       1       2       3
```

Element wybieramy przez `list[index]`. Ujemne indeksy liczą od końca: `-1` to ostatni.


````python
names = ["Anna", "Tom", "Kate", "John"]

print(names[0])    # Anna  — the first element
print(names[2])    # Kate
print(names[-1])   # John  — the last element
print(names[-2])   # Kate  — second from the end
````

⚠️ Odwołanie do nieistniejącego indeksu (np. `names[10]`) kończy się błędem `IndexError`. Największy poprawny indeks to `len(list) - 1`.

### 1.2 Wycinki (slicing) — fragment listy

`list[start:stop]` zwraca **nową** listę od `start` do `stop` (bez `stop`), dokładnie jak w `range()`.


````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[1:4])    # [20, 30, 40]   — indices 1, 2, 3
print(numbers[:3])     # [10, 20, 30]   — from the start
print(numbers[3:])     # [40, 50, 60]   — to the end
print(numbers[::2])    # [10, 30, 50]   — every second element
````

## Część 2 — Modyfikowanie listy ✏️

Listy są **zmienne** (*mutable*) — po utworzeniu można je swobodnie przebudowywać.

| Metoda | Działanie |
|--------|-----------|
| `list.append(x)` | dodaj `x` na **koniec** |
| `list.insert(i, x)` | wstaw `x` na pozycję `i` |
| `list.remove(x)` | usuń **pierwsze** wystąpienie `x` |
| `list.pop()` | usuń i zwróć **ostatni** element |
| `list.pop(i)` | usuń i zwróć element o indeksie `i` |
| `list[i] = x` | podmień element na pozycji `i` |


````python
shopping = ["bread", "milk"]
print("Start:    ", shopping)

shopping.append("eggs")          # add to the end
print("Append:   ", shopping)

shopping.insert(0, "butter")     # put at the front
print("Insert:   ", shopping)

shopping.remove("milk")          # remove by value
print("Remove:   ", shopping)

last = shopping.pop()            # remove and keep the last one
print("Popped:   ", last, "->", shopping)

shopping[0] = "margarine"        # replace by index
print("Replaced: ", shopping)
````

## Część 3 — Przydatne operacje 🧰

Wiele rzeczy Python policzy za nas:

| Operacja | Wynik |
|----------|-------|
| `len(list)` | liczba elementów |
| `x in list` | `True`/`False` — czy jest w liście |
| `list1 + list2` | połączenie dwóch list |
| `sum(list)` | suma (liczby) |
| `min(list)` / `max(list)` | najmniejszy / największy |
| `sorted(list)` | **nowa** posortowana lista |
| `list.sort()` | sortuje **w miejscu** |
| `list.reverse()` | odwraca **w miejscu** |


````python
marks = [5, 3, 4, 6, 2]

print("Count:   ", len(marks))
print("Sum:     ", sum(marks))
print("Highest: ", max(marks))
print("Lowest:  ", min(marks))
print("Average: ", sum(marks) / len(marks))
print("Sorted:  ", sorted(marks))       # a new sorted list
print("Original:", marks)               # unchanged by sorted()

marks.sort()                            # sorts in place
print("In place:", marks)
````

💡 Różnica: `sorted(list)` **zwraca nową** listę i nie rusza oryginału, a `list.sort()` **zmienia** oryginał i nic nie zwraca. To częste źródło pomyłek.

## Część 4 — Iteracja (przechodzenie po liście) 🔁

Najczęstsza operacja: zrobić coś z **każdym** elementem. Pętla `for` chodzi po liście bezpośrednio — bez indeksów.


````python
names = ["Anna", "Tom", "Kate"]

for name in names:
    print("Hello,", name)
````

### 4.1 Iteracja z obliczeniem

Zmienną pętli możemy użyć w działaniu — np. podnieść każdą cenę o VAT.


````python
prices = [100, 250, 40]

for price in prices:
    with_vat = price * 1.23
    print(price, "->", round(with_vat, 2), "with VAT")
````

### 4.2 `enumerate` — element **i** jego indeks

Gdy potrzebujesz numeru pozycji razem z wartością, użyj `enumerate()`. Zwraca parę `(index, value)`.


````python
names = ["Anna", "Tom", "Kate"]

for index, name in enumerate(names):
    print(index, "->", name)

# Often nicer to start counting from 1:
for position, name in enumerate(names, start=1):
    print("Runner", position, "is", name)
````

## Część 5 — Wyszukiwanie 🔎

### 5.1 Czy element istnieje — operator `in`

Najprostsze pytanie: *czy X jest na liście?* Odpowiada operator `in` (zwraca `True`/`False`).


````python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)     # True
print("mango" in fruits)      # False

# Typical use inside an if:
wanted = "cherry"
if wanted in fruits:
    print("Yes, we have", wanted)
else:
    print("Sorry, no", wanted)
````

### 5.2 Gdzie jest element — `index` i `count`

- `list.index(x)` — zwraca **indeks** pierwszego wystąpienia `x` (błąd, jeśli go nie ma!),
- `list.count(x)` — ile razy `x` występuje.


````python
letters = ["a", "b", "c", "b", "a", "b"]

print("Position of first 'c':", letters.index("c"))   # 2
print("Number of 'b':        ", letters.count("b"))   # 3

# Safe search: check first, then ask for the index
target = "z"
if target in letters:
    print("Found at", letters.index(target))
else:
    print(target, "is not in the list")
````

### 5.3 Wyszukiwanie własną pętlą — np. najdłuższe słowo

Gotowe funkcje nie zawsze wystarczą. Często szukamy elementu spełniającego **warunek** — robimy to pętlą, pamiętając „najlepszego dotychczas".


````python
words = ["cat", "elephant", "dog", "butterfly", "ant"]

longest = words[0]                 # assume the first is the longest
for word in words:
    if len(word) > len(longest):   # found a longer one?
        longest = word             # remember it

print("The longest word is:", longest)
````

## Część 6 — Filtrowanie 🧹

**Filtrowanie** = utworzenie **nowej** listy zawierającej tylko te elementy, które spełniają warunek.

### 6.1 Sposób klasyczny — pusta lista + `for` + `if`

Schemat, który warto zapamiętać: zaczynamy od pustej listy i **dokładamy** pasujące elementy.


````python
numbers = [4, 7, 10, 3, 18, 9, 22, 1]

even_numbers = []                 # start empty
for n in numbers:
    if n % 2 == 0:                # condition: is it even?
        even_numbers.append(n)    # keep it

print("All numbers: ", numbers)
print("Even only:   ", even_numbers)
````

### 6.2 List comprehension — ten sam efekt w jednej linii

Python ma elegancki skrót na „nowa lista z elementów spełniających warunek":

```python
new_list = [expression for item in list if condition]
```

Czytaj: *„weź `item` z listy, **jeśli** `condition`, i umieść `expression` w nowej liście"*.


````python
numbers = [4, 7, 10, 3, 18, 9, 22, 1]

# Same result as the loop above, in one line:
even_numbers = [n for n in numbers if n % 2 == 0]
print("Even:", even_numbers)

# The expression can transform items, e.g. square the odd ones:
odd_squared = [n ** 2 for n in numbers if n % 2 == 1]
print("Odd squared:", odd_squared)

# Filtering text — names longer than 3 letters:
names = ["Jo", "Anna", "Tom", "Kate", "Ed"]
long_names = [name for name in names if len(name) > 3]
print("Long names:", long_names)
````

💡 Oba sposoby są **poprawne**. Pętla z `if` jest czytelniejsza na start; *list comprehension* jest zwięzłe i bardzo „pythonowe". Na egzaminie/kolokwium warto umieć oba.

### 6.3 Przykładowy program — analiza ocen

Łączymy iterację, wyszukiwanie i filtrowanie w jednym miejscu.


````python
marks = [5, 2, 4, 1, 6, 3, 2, 5, 4]

passed = [m for m in marks if m >= 3]     # filter: positive marks
failed = [m for m in marks if m < 3]      # filter: negative marks

print("All marks:     ", marks)
print("Passed:        ", passed)
print("Failed:        ", failed)
print("How many passed:", len(passed))
print("Best mark:     ", max(marks))
print("Average:       ", round(sum(marks) / len(marks), 2))
````

## 📝 Ściąga — listy

```python
# --- CREATE & ACCESS ---
my_list = [1, 2, 3]
my_list[0]            # first element (indexing from 0)
my_list[-1]           # last element
my_list[1:3]          # slice (new list)
len(my_list)          # length

# --- MODIFY ---
my_list.append(x)     # add at the end
my_list.insert(i, x)  # insert at position i
my_list.remove(x)     # remove first x
my_list.pop()         # remove & return last
my_list[i] = x        # replace by index

# --- USEFUL ---
sum(lst)  min(lst)  max(lst)
sorted(lst)           # new sorted list
lst.sort()            # sort in place
x in lst              # membership test

# --- ITERATE ---
for item in lst:
    ...
for i, item in enumerate(lst, start=1):
    ...

# --- SEARCH ---
lst.index(x)          # position of first x
lst.count(x)          # how many x

# --- FILTER ---
result = []
for item in lst:
    if condition:
        result.append(item)
# or, equivalently:
result = [item for item in lst if condition]
```
