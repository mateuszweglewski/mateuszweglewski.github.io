## Część 1 — Po co nam funkcje 🧱

Wyobraź sobie, że w kilku miejscach programu musisz przywitać użytkownika. Bez funkcji kopiujesz te same linijki wielokrotnie — a gdy coś trzeba zmienić, poprawiasz w wielu miejscach.

**Funkcja** to nazwany blok kodu, który:
- piszemy **raz**,
- **wywołujemy** (uruchamiamy) **wiele razy**,
- opcjonalnie przyjmuje dane (**parametry**) i oddaje wynik (**`return`**).

Funkcje `print()`, `len()`, `input()`, `range()` znasz już od pierwszych zajęć — to funkcje **wbudowane**. Teraz napiszemy **własne**.

### 1.1 Definicja funkcji — `def`

```python
def function_name():
    # body — indented, just like in a loop
    ...
```

Dwie zasady (jak przy `if`/`for`): po nagłówku stawiamy **dwukropek `:`**, a ciało funkcji **wcinamy**.

**Uwaga:** samo zdefiniowanie funkcji **nic nie wypisuje** — kod uruchomi się dopiero, gdy funkcję **wywołamy** przez `function_name()`.


````python
# Define the function
def greet():
    print("Hello!")
    print("Welcome to the lecture.")

# Nothing happened yet — we only defined it.
# Now let's CALL it (as many times as we like):
greet()
greet()
````

## Część 2 — Parametry (dane wejściowe) 📥

Funkcja bez danych jest mało elastyczna. **Parametry** to zmienne w nawiasie, przez które przekazujemy funkcji informacje.

```python
def greet(name):        # 'name' is a parameter
    print("Hello,", name)

greet("Anna")           # "Anna" is an argument
greet("Tom")
```

**Pojęcia:** *parametr* to nazwa w definicji, *argument* to konkretna wartość przy wywołaniu. W praktyce używa się tych słów zamiennie.


````python
def greet(name):
    print("Hello,", name, "!")

greet("Anna")
greet("Tom")
greet("Kate")
````

### 2.1 Wiele parametrów

Parametrów może być więcej — oddzielamy je przecinkami. **Kolejność ma znaczenie**: pierwszy argument trafia do pierwszego parametru itd.


````python
def introduce(name, age):
    print(name, "is", age, "years old.")

introduce("Anna", 20)
introduce("Tom", 17)
````

### 2.2 Parametr może być dowolnego typu — także lista

Do funkcji można przekazać liczbę, tekst, wartość logiczną, a także całą **listę**.


````python
def print_all(items):
    for item in items:
        print("-", item)

shopping = ["bread", "milk", "eggs"]
print_all(shopping)
````

## Część 3 — `return`, czyli zwracanie wyniku 📤

Dotąd funkcje tylko **wypisywały** (`print`). Znacznie częściej chcemy, aby funkcja **obliczyła wynik i oddała go** z powrotem — żebyśmy mogli go dalej użyć. Służy do tego słowo **`return`**.

```python
def add(a, b):
    return a + b

result = add(2, 3)     # the function gives back 5
print(result)          # 5
```

**`print` vs `return` — kluczowa różnica!**
- `print` tylko **pokazuje** tekst na ekranie i znika.
- `return` **oddaje wartość** do programu — można ją zapisać w zmiennej, dodać, przekazać dalej.


````python
def add(a, b):
    return a + b

# The returned value can be stored and reused:
total = add(2, 3)
print("Sum is:", total)

# ...or used directly in a bigger expression:
print("Double the sum:", add(2, 3) * 2)
````

### 3.1 `return` kończy działanie funkcji

W momencie `return` funkcja **natychmiast się kończy** — kod po nim już się nie wykona.


````python
def absolute_value(n):
    if n >= 0:
        return n        # if positive, return here and stop
    return -n           # this runs only for negative n

print(absolute_value(5))    # 5
print(absolute_value(-8))   # 8
````

### 3.2 Funkcja bez `return`

Jeśli funkcja nie ma `return`, zwraca specjalną wartość `None` („nic"). To typowy błąd początkujących: `print` w środku to **nie** to samo co `return`.


````python
def add_wrong(a, b):
    print(a + b)        # only prints — returns nothing!

x = add_wrong(2, 3)     # prints 5...
print("x is:", x)       # ...but x is None
````

### 3.3 Przykład — funkcja oceniająca

`return` świetnie łączy się z `if`. Funkcja może zwrócić np. tekstową ocenę na podstawie liczby punktów.


````python
def grade(points):
    if points >= 90:
        return "A"
    elif points >= 75:
        return "B"
    elif points >= 50:
        return "C"
    else:
        return "F"

print("95 ->", grade(95))
print("60 ->", grade(60))
print("30 ->", grade(30))
````

## Część 4 — Rodzaje argumentów 🎛️

### 4.1 Argumenty domyślne

Parametrowi można nadać **wartość domyślną** — użyta, gdy wywołujący jej nie poda.


````python
def greet(name, greeting="Hello"):
    print(greeting + ",", name)

greet("Anna")                 # uses the default -> Hello, Anna
greet("Tom", "Good morning")  # overrides it     -> Good morning, Tom
````

### 4.2 Argumenty nazwane (keyword arguments)

Przy wywołaniu można podać argumenty **po nazwie** — wtedy kolejność nie ma znaczenia i kod jest czytelniejszy.


````python
def describe_pet(name, species, age):
    print(name, "is a", species, "aged", age)

# Positional (order matters):
describe_pet("Rex", "dog", 3)

# By name (order does not matter, very readable):
describe_pet(species="cat", age=2, name="Whiskers")
````

### 4.3 Funkcja zwracająca kilka wartości

`return` może oddać kilka wartości naraz (oddzielonych przecinkami). Odbieramy je do kilku zmiennych.


````python
def min_and_max(numbers):
    return min(numbers), max(numbers)

lowest, highest = min_and_max([4, 9, 1, 7, 3])
print("Lowest:", lowest)
print("Highest:", highest)
````

## Część 5 — Zakres zmiennych (scope) 🔒

Zmienna utworzona **wewnątrz** funkcji jest **lokalna** — istnieje tylko w czasie działania tej funkcji i **nie jest widoczna na zewnątrz**.


````python
def calculate():
    result = 42          # local variable — lives only inside calculate()
    print("Inside:", result)

calculate()
# print(result)   # <- this would raise a NameError: 'result' is not defined
````

Dlatego **właściwym sposobem** wyjęcia wyniku z funkcji jest `return`, a **nie** poleganie na zmiennych wewnętrznych.

Zmienne zdefiniowane **poza** funkcjami są **globalne** — funkcja może je odczytać, ale dobrą praktyką jest przekazywać dane przez **parametry**, a wynik oddawać przez **`return`**. Kod jest wtedy przewidywalny i łatwy do testowania.


````python
tax_rate = 0.23        # global — read-only guideline inside functions

def gross_price(net):
    # we READ the global tax_rate, but take 'net' as a parameter
    return net * (1 + tax_rate)

print("Gross of 100:", gross_price(100))
````

### 5.1 Składanie funkcji

Prawdziwa siła funkcji: jedna może **wywoływać** inną. Budujemy program z małych, czytelnych klocków.


````python
def square(n):
    return n * n

def sum_of_squares(a, b):
    return square(a) + square(b)      # reuse square() twice

print("3^2 + 4^2 =", sum_of_squares(3, 4))   # 25
````

## 📝 Ściąga — funkcje

```python
# --- DEFINE & CALL ---
def function_name(param1, param2):
    # body (indented)
    result = param1 + param2
    return result            # give the value back

x = function_name(2, 3)      # call and store the result

# --- PARAMETERS ---
def greet(name):             # one parameter
    print("Hi,", name)

def greet(name, greeting="Hello"):   # default argument
    print(greeting, name)

greet(name="Anna")           # keyword argument (order-free)

# --- RETURN ---
return value                 # give one value back, end the function
return a, b                  # return several values at once
# no return  ->  function returns None

# --- SCOPE ---
# variables created inside a function are LOCAL (invisible outside)
# pass data IN via parameters, get data OUT via return
```

### Najczęstsze błędy
- **`print` zamiast `return`** — funkcja wtedy nic nie oddaje (zwraca `None`).
- **Zapomniane wywołanie** — sama definicja (`def`) nic nie uruchamia.
- **Oczekiwanie dostępu do zmiennej lokalnej z zewnątrz** — nie zadziała; użyj `return`.
